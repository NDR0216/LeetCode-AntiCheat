```python
from collections import defaultdict
import csv
import json
import os
import shutil
from tqdm import tqdm

## --- PARAMETER ---
CONTEST = "BC188" # Contest name
variable_Q2 = "velmoritha"
variable_Q3 = "norvelithx"
variable_Q4 = "telmorvian"
```

```python
cheater = defaultdict(list)

for i in range(12):
    for filename in tqdm(os.listdir(CONTEST+"/Q2_replay/"+str(i)+"000-"+str(i)+"999")):
        f_path = os.path.join(CONTEST+"/Q2_replay/"+str(i)+"000-"+str(i)+"999", filename)
        if os.path.isfile(f_path):
            with open(f_path, "r") as f:
                if variable_Q2 in f.read():
                    rank = int(filename[:-4])
                    
                    cheater[rank].append(2)
                    
for i in range(12):
    for filename in tqdm(os.listdir(CONTEST+"/Q3_replay/"+str(i)+"000-"+str(i)+"999")):
        f_path = os.path.join(CONTEST+"/Q3_replay/"+str(i)+"000-"+str(i)+"999", filename)
        if os.path.isfile(f_path):
            with open(f_path, "r") as f:
                if variable_Q3 in f.read():
                    rank = int(filename[:-4])
                    
                    cheater[rank].append(3)

for i in range(12):
    for filename in tqdm(os.listdir(CONTEST+"/Q4_replay/"+str(i)+"000-"+str(i)+"999")):
        f_path = os.path.join(CONTEST+"/Q4_replay/"+str(i)+"000-"+str(i)+"999", filename)
        if os.path.isfile(f_path):
            with open(f_path, "r") as f:
                if variable_Q4 in f.read():
                    rank = int(filename[:-4])
                    
                    cheater[rank].append(4)

print(cheater)
```

```python
## record the rank, user name, and submission ID of the mass cheating group to csv

## --- PARAMETER ---
CSV = CONTEST+"/replay.csv" # the name of csv gonna be created to record the mass cheating group


print("how many cheater:", len(cheater))

with open(CONTEST+"/"+CONTEST+".csv") as f:
    rows = list(csv.reader(f))
    
    with open(CSV, "w") as csv_:
        print("rank,user,Q2,Q3,Q4", file=csv_)
        
        for i in sorted(cheater):
            Q2 = ""
            if 2 in cheater[i]:
                Q2 = rows[i][2]
            Q3 = ""
            if 3 in cheater[i]:
                Q3 = rows[i][3]
            Q4 = ""
            if 4 in cheater[i]:
                Q4 = rows[i][4]
            
            print(str(i)+"," + rows[i][0]+"," + Q2+"," + Q3+"," + Q4, file=csv_)
```
